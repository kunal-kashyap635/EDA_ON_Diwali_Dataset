# EDA

```python
df.columns
```

# Gender wise distribution

- Insights :- Most of the buyers are females and female purchasing power is greater than males

```python
ax = sns.countplot(data=df, x='Gender', hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_per_gender = df.groupby('Gender', as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
```

```python
sales_per_gender
```

```python
ax = sns.barplot(data=sales_per_gender, x='Gender', y='Amount',hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars, fmt='%.0f')
```

# Age Wise Distribution

- Insights : Most of the buyers are from Age Group 26-35 and mostly are females.

```python
ax = sns.countplot(data=df, x='Age Group', hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_per_age = df.groupby('Age Group', as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

ax = sns.barplot(data=sales_per_age, x='Age Group', y='Amount', hue='Age Group')
for bars in ax.containers:
    ax.bar_label(bars, fmt='%.0f')
```

# State Wise Total Orders

- Insights : Most of the orders are from Uttar Pradesh, Maharashtra and Karnataka.

```python
state_wise_order = df.groupby('State', as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False)
```

```python
plt.figure(figsize=(30,7))
ax = sns.barplot(data=state_wise_order, x='State', y='Orders', hue='State')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
state_wise_amount = df.groupby('State', as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
```

```python
plt.figure(figsize=(30,7))
ax = sns.barplot(data=state_wise_amount, x='State', y='Amount', hue='State')
for bars in ax.containers:
    ax.bar_label(bars, fmt='%.0f')
```

# Marital Status

- Insights : Most of the buyers are married women and they have high purchasing power

```python
ax = sns.countplot(data=df, x='Marital_Status', hue='Marital_Status')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
marital_status_amount = df.groupby(['Marital_Status','Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

ax = sns.barplot(data=marital_status_amount, x='Marital_Status', y='Amount',hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars, fmt='%.0f')
```

# Occupation wise Distribution

- Insights : Most of the buyers are from IT Sector, Aviation and Healtcare sector.

```python
plt.figure(figsize=(30,7))
ax = sns.countplot(data=df, x='Occupation', hue='Occupation')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
occupation_amount = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

plt.figure(figsize=(30,7))
sns.barplot(data=occupation_amount, x='Occupation', y='Amount',hue='Occupation')
```

# Product Category

- Insights - Most sold products are from food, clothing and Electronics

```python
plt.figure(figsize=(30,7))
ax = sns.countplot(data=df, x='Product_Category', hue='Product_Category')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
prd_cat_amount = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

plt.figure(figsize=(32,7))
sns.barplot(data=prd_cat_amount, x='Product_Category', y='Amount',hue='Product_Category')
```

```python
prd_ord = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

plt.figure(figsize=(32,7))
ax = sns.barplot(data=prd_ord, x='Product_ID', y='Orders',hue='Product_ID')
for bars in ax.containers:
    ax.bar_label(bars)
```
